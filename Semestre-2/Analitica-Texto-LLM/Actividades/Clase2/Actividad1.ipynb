## Procesamiento del Lenguaje

NLTK permite realizar búsquedas y comparaciones rápidas dentro de un texto:
* **Concordance:** Buscar una palabra en un texto y ver su contexto.
* **Similar:** Buscar palabras similares a la indicada.
* **Common contexts:** Examinar el contexto compartido por 2 o más palabras.
* **Len:** Obtener la longitud de un texto (cantidad total de tokens/palabras).

```python
# Buscar la palabra "monstrous" en text1
text1.concordance("monstrous") 
```

```python
# Buscar palabras similares a "monstrous"
text1.similar("monstrous") 
text2.similar("monstrous") 
```

```python
# Contexto compartido
text2.common_contexts(["monstrous", "very"]) 
```

## Vocabulario, Riqueza Léxica y Frecuencia

Se puede calcular el vocabulario único de un texto utilizando `set()` y ordenarlo con `sorted()`. Dividiendo el vocabulario único entre el total de palabras, se obtiene la **riqueza léxica**. Además, se puede contar la frecuencia de palabras específicas.

```python
# Calcular el vocabulario único de un texto
vocabulario_text3 = sorted(set(text3)) 
print("Vocavulario en text3:", vocabulario_text3)
print("Palabras únicas en text3:", len(set(text3)))
```

```python
# Riqueza léxica de un texto
riqueza_lexica = len(set(text3)) / len(text3) 
print("Riqueza léxica de text3:", riqueza_lexica)
```

```python
# Calcular frecuencia de una palabra
print("Apariciones de 'smote' en text3:", text3.count("smote"))
```

```python
# Porcentaje de aparición de 'a' en text4
porcentaje_a = 100 * text4.count('a') / len(text4) 
print("Porcentaje de 'a' en text4:", porcentaje_a)
```

## Acceso a corpora de texto y recursos léxicos

NLTK incluye acceso a corpus extensos como el Proyecto Gutenberg. Se puede iterar sobre sus archivos para obtener estadísticas básicas de caracteres, palabras, oraciones y vocabulario.

```python
# El modelo preentrenado Punkt ayuda asaber dónde termina una oración y empieza otra 
# (para distinguir, por ejemplo, si un punto es un final de oración o una abreviatura)
import nltk
nltk.download('punkt_tab')
```

```python
from nltk.corpus import gutenberg 

# Ver los archivos disponibles
print(gutenberg.fileids()) 

# Acceder a las palabras de un texto específico
emma = gutenberg.words('austen-emma.txt') 

# Estadísticas de todos los archivos del corpus Gutenberg
for fileid in gutenberg.fileids(): 
    num_chars = len(gutenberg.raw(fileid)) 
    num_words = len(gutenberg.words(fileid)) 
    num_sents = len(gutenberg.sents(fileid)) 
    num_vocab = len(set([w.lower() for w in gutenberg.words(fileid)])) 
    
    # Imprimir promedios (chars/word, words/sent, words/vocab)
    print(round(num_chars/num_words), round(num_words/num_sents), round(num_words/num_vocab), fileid)
```

## Funciones personalizadas y Recursos Léxicos

Se pueden definir funciones que pueden recibir un parámetro y retornar un valor. Tambien se puede acceder a diccionarios de pronunciación, listas de palabras vacías (stopwords) y WordNet, una enorme base de datos léxica que agrupa palabras en conjuntos de sinónimos llamados *synsets*.

```python
# Definir una función de riqueza léxica
def lexical_diversity(text): 
    value = len(text) / len(set(text)) 
    return value

from nltk.corpus import genesis 
kjv = genesis.words('english-kjv.txt') 
print("Diversidad del corpus genesis:", lexical_diversity(kjv))
```

```python
# Palabras vacías (Stopwords)
from nltk.corpus import stopwords 
print("Stopwords en inglés (primeras 10):", stopwords.words('english')[:10])
```

```python
# Diccionario de pronunciación
entries = nltk.corpus.cmudict.entries() 
for entry in entries[42371:42379]: 
    print(entry) 
```

```python
# Lista de palabras comparativas
from nltk.corpus import swadesh
print(swadesh.fileids())

fr2en = swadesh.entries(['fr','en'])
print(fr2en)
```

```python
# Uso de Wordnet
from nltk.corpus import wordnet as wn 
print("Synsets de motorcar:", wn.synsets('motorcar')) 
print("Lemas de car.n.01:", wn.synset('car.n.01').lemmas()) 
print("Definición de car.n.01:", wn.synset('car.n.01').definition())
```

## Procesamiento de texto plano

Acceder a archivos locales, dividir los textos en oraciones, aplicar *Stemming* (encontrar la raíz) y *Lematización* (encontrar la forma base de diccionario).

*(Nota: El código para abrir `moby.txt` fallará a menos que se tenga un archivo de texto con ese nombre en la carpeta local).*

```python
# Acceder a archivos locales (Ejemplo)
f = open('moby.txt') 
raw = f.read() 
tokens = nltk.word_tokenize(raw)
```

```python
# Revisar los tokens
type(tokens)
print("La cantidad de tokens son:", len(tokens))
print("Los 10 primeros tokens son;", tokens[:10])
```

```python
# Diferentes formas de la misma palabra
input1 = "List listed lists listing listings" 
words1 = input1.lower().split() 

udhr = nltk.corpus.udhr.words('English-Latin1')
print(udhr)
```

```python
# Encontrar raíz de palabra (Stemming)
porter = nltk.PorterStemmer() 
print("Stemming:", [porter.stem(t) for t in words1])
print("Stemming:", [porter.stem(t) for t in udhr[:20]])
```

```python
# Lematización
WNlemma = nltk.WordNetLemmatizer() 
print("Lemmatization:", [WNlemma.lemmatize(t) for t in udhr[:20]])
```

```python
# Dividir en oraciones
texto = "This is the first sentence. A gallon of milk in the U.S. costs $2.99. Is this the third sentence? Yes, it is!" 
sentences = nltk.sent_tokenize(texto) 
print("Oraciones detectadas:", sentences)
```

## Categorización y Etiquetado POS (Part-of-Speech)

Identificar la etiqueta POS de una lista de palabras. Un token etiquetado en NLTK se representa como una tupla. También se puede revisar la frecuencia de las etiquetas.

```python
# Ayuda del UPENN Tagset
nltk.download('tagsets')
nltk.download('tagsets_json')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
nltk.help.upenn_tagset('MD')
```

```python
# Identificar etiqueta POS
text = nltk.word_tokenize("And now for something completely different") 
print("POS Tagging:", nltk.pos_tag(text))
```

```python
# Ambigüedad
text2 = nltk.word_tokenize("Visiting aunts can be a nuisance") 
print("POS Tagging con ambigüedad:", nltk.pos_tag(text2))
```

```python
# Tuplas etiquetadas
tagged_token = nltk.tag.str2tuple('fly/NN') 
print("Tupla:", tagged_token)
```

```python
# Conjunto de etiquetas universal y Frecuencias
from nltk.corpus import brown 
brown_news_tagged = brown.tagged_words(categories='news', tagset='universal') 
tag_fd = nltk.FreqDist(tag for (word, tag) in brown_news_tagged) 
print("Etiquetas más comunes:", tag_fd.most_common()[:5])
```

## Análisis de la estructura de una oración

Dar sentido a las oraciones es fácil si siguen una estructura gramatical bien definida. A continuación, se procede a usar una Gramática Libre de Contexto (CFG) para analizar "Alice loves Bob".

```python
texto = nltk.word_tokenize("Alice loves Bob") 

# Definimos la gramática
grammar = nltk.CFG.fromstring("""
S -> NP VP
VP -> V NP
NP -> 'Alice' | 'Bob'
V -> 'loves'
""") 

parser = nltk.ChartParser(grammar) 
trees = parser.parse_all(texto) 

# Imprimir el árbol sintáctico
for tree in trees: 
    print(tree)
```
